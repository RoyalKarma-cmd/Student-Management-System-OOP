```python
class Student:

    def __init__(self, student_id, name, age, branch, marks):
        self.student_id = student_id
        self.name = name
        self.age = age
        self.branch = branch
        self.grades = marks

    def show_details(self):
        print("\nStudent_ID:", self.student_id)
        print("Name:", self.name)
        print("Age:",self.age)
        print("Branch:",self.branch)
        print("Grades:",self.grades)

    def greet(self):
        print(f"\nWelcome aboard {self.name}")
        print(f"Welcome to the {self.branch} Department")

    def calculate_grades(self):
        if self.grades >=90:
            print("\nGrades: A+")
        elif self.grades >=80:
            print("\nGrades: A")
        elif self.grades >= 70:
            print("\nGrades: B")
        elif self.grades >= 60:
            print("\nGrades: C")
        elif self.grades >= 50:
            print("\nGrades: D")
        else:
            print("\nGrades: F")


student1 = Student(100, "Jaydeep", 21, "Extc", 92)
student2 = Student(101, "Bhumi", 21, "Extc", 88)
student3 = Student(102, "Saish", 16, "Merchant Navy", 90)
student4 = Student(103, "Aditya", 21, "Cse", 86)
student5 = Student(104, "Aarya", 21, "Cse", 79)
student6 = Student(105, "Sarvesh", 21, "Electrical", 92)
student7 = Student(106, "Swayam", 20, "Extc", 75)
student8 = Student(107, "Shantanu", 24, "Extc", 68)
student9 = Student(108, "Sayali", 26, "Extc", 95)


students = [
    student1,
    student2,
    student3,
    student4,
    student5,
    student6,
    student7,
    student8,
    student9
]

for student in students:
    student.show_details()
    student.greet()
    student.calculate_grades()
```
